```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from mpl_toolkits.mplot3d import Axes3D

# Sample data
area = np.array([1000, 1200, 1500, 1800, 2000, 2200, 2500, 2800])
bedrooms = np.array([2, 2, 3, 3, 4, 4, 5, 5])
price = np.array([50, 60, 75, 90, 110, 125, 145, 170])

# Create input matrix
X = np.column_stack((area, bedrooms))
y = price

# Create and train the regression model
model = LinearRegression()
model.fit(X, y)

# Predictions
y_pred = model.predict(X)
# Create a mesh grid for the regression plane
area_grid, bedroom_grid = np.meshgrid(
    np.linspace(area.min(), area.max(), 20),
    np.linspace(bedrooms.min(), bedrooms.max(), 20)
)

# Predict prices for the grid
price_grid = model.predict(
    np.column_stack((area_grid.ravel(), bedroom_grid.ravel()))
).reshape(area_grid.shape)

# Create 3D plot
fig = plt.figure(figsize=(10, 7))
ax = fig.add_subplot(111, projection='3d')

# Plot actual data points
ax.scatter(area, bedrooms, price, color='blue',
           s=60, label='Actual House Prices')

# Plot regression plane
ax.plot_surface(area_grid, bedroom_grid, price_grid,alpha=0.5, color='orange')

# Labels
ax.set_xlabel('Area (sq ft)')
ax.set_ylabel('Number of Bedrooms')
ax.set_zlabel('House Price (lakhs)')

ax.set_title('3D Linear Regression: House Price Prediction')

ax.legend()

plt.show()
# Display model coefficients
print("Intercept:", model.intercept_)
print("Coefficients:")
print("Area:", model.coef_[0])
print("Bedrooms:", model.coef_[1])
```

**What the 3D plot shows**

The three axes represent:

X-axis: Area in square feet

Y-axis: Number of bedrooms

Z-axis: House price


The blue dots are the actual house-price observations.

The orange surface is the regression plane produced by the model. It represents the house prices predicted from Area + Number of Bedrooms.

The regression equation is:
House Price=b0+b1(Area)+b2(Bedrooms)

So this is a multiple linear regression model, visualized in 3D.
