```python
import matplotlib.pyplot as plt
from matplotlib.patches import Circle

# Layers and neurons
input_features = [
    "TransactionAmount",
    "TransactionTime",
    "MerchantCategory",
    "CustomerAge",
    "AccountBalance",
    "NumberOfTransactionsToday"
]

hidden_layer = ["Neuron 1", "Neuron 2", "Neuron 3", "Neuron 4"]
output_layer = ["Fraud\n(0 = Genuine,\n1 = Fraud)"]

# Positions
layers = [
    (0, input_features),
    (1, hidden_layer),
    (2, output_layer)
]

fig, ax = plt.subplots(figsize=(12, 8))

# Draw neurons
positions = {}

for x, neurons in layers:
    n = len(neurons)

    for i, neuron in enumerate(neurons):
        y = (n - 1) / 2 - i
        positions[(x, i)] = (x, y)

        circle = Circle((x, y), 0.18, fill=False, linewidth=2)
        ax.add_patch(circle)

        ax.text(x, y, neuron, ha="center", va="center", fontsize=9)

# Connect input layer → hidden layer
for i in range(len(input_features)):
    for j in range(len(hidden_layer)):
        x1, y1 = positions[(0, i)]
        x2, y2 = positions[(1, j)]

        ax.plot([x1, x2], [y1, y2], linewidth=0.7)

# Connect hidden layer → output layer
for i in range(len(hidden_layer)):
    x1, y1 = positions[(1, i)]
    x2, y2 = positions[(2, 0)]

    ax.plot([x1, x2], [y1, y2], linewidth=0.7)

# Layer labels
ax.text(0, 3.3, "INPUT LAYER", ha="center",
        fontsize=13, fontweight="bold")

ax.text(1, 3.3, "HIDDEN LAYER", ha="center",
        fontsize=13, fontweight="bold")

ax.text(2, 3.3, "OUTPUT LAYER", ha="center",
        fontsize=13, fontweight="bold")

# Title
ax.set_title(
    "Neural Network Architecture for Credit Card Fraud Detection",
    fontsize=16,
    fontweight="bold"
)

# Formatting
ax.set_xlim(-0.5, 2.5)
ax.set_ylim(-3.5, 3.7)
ax.axis("off")

plt.show()
```
